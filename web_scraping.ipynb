```python
from bs4 import BeautifulSoup
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys 
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
```

```python
#Set url and initialize selenium webdriver
url = "https://www.yellowpages.co.th/"
options = Options()

#dont open a visible browser window
# options.add_argument("--headless")

#Set up Chrome driver
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)

#Open the webpage and maximize window
driver.maximize_window()
driver.get(url)
```

```python
#Create wait object to wait for elements to load
wait = WebDriverWait(driver, 10)

#Find element CLASS_NAME for search then sent keys to search for "ร้านอาหาร" and press Enter
wait.until(EC.presence_of_element_located((By.CLASS_NAME, "input-text-what"))).send_keys("ร้านอาหาร", Keys.ENTER)
```

---
Save detail business each pages 
---

- Business name
- Website
- address
- category
- description

```python
#Initialize page counter 
num_pages = 1

# Wait for results to load
time.sleep(3)

# Store ALL data
all_data = []

#Set number of pages to scrape(set 10 for testing)
page_items = 10
for page in range(page_items):
    
    # Get all listing elements on the page
    listings = driver.find_elements(By.CLASS_NAME, "yp-search-listing")

    for listing in listings:
        try:
            #Get Business name
            name = listing.find_element(By.CSS_SELECTOR, "h3 a").text
        except:
            name = "N/A"
        
        try:
            #Get Business website
            website = listing.find_element(By.CSS_SELECTOR, "a.yp-listing-website").text
        except:
            website = "N/A"
        
        try:
            #Get Business address
            address = listing.find_element(By.CLASS_NAME, "yp-listing-address").text
        except:
            address = "N/A"
        
        try:
            #Get Business description
            description = listing.find_element(By.CLASS_NAME, "yp-listing-desc").text
        except:
            description = "N/A"
        
        try:
            #Get Business category
            category = listing.find_element(By.CSS_SELECTOR, ".listing-category-section a").text.strip()
        except:
            category = "N/A"
        
        all_data.append({
            "name": name,
            "website": website,
            "address": address,
            "category": category,
            "description": description
        })
    
    
    # Go to next page - Use "Next" button instead!
    try:
        # Scroll to pagination first
        pagination = driver.find_element(By.CSS_SELECTOR, "ul.pagination")
        driver.execute_script("arguments[0].scrollIntoView();", pagination)
        time.sleep(1)
        
        # Click "Next" button to go to next page
        next_btn = driver.find_element(By.CSS_SELECTOR, "li.pager__item--next a")
        next_btn.click()
        time.sleep(3)
    except Exception as e:
        print(f"Finished! Error: {e}")
        break
```

```python

# Convert to DataFrame and display first 10 rows
df = pd.DataFrame(all_data)
df.head(10)
```

```python
# Save DataFrame to CSV file with UTF-8 encoding
df.to_csv('business_data.csv', index=False, encoding='utf-8-sig')
```
